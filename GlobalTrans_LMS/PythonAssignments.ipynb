```python
def division(a, b):
    try:
        # Convert to integers INSIDE the try block so errors are caught
        num1 = int(a)
        num2 = int(b)
        out_list.append(num1 // num2)
    except ZeroDivisionError as i:
        out_list.append("Error Code: " + str(i))
    except ValueError as j:
        out_list.append("Error Code: " + str(j))
        
out_list = []
T = int(input())
for i in range(T):
    temp_str = input()
    temp_list = temp_str.split()
    
    # Check if the user actually entered enough values to avoid an IndexError
    if len(temp_list) >= 2:
        # Pass the raw string inputs to the function
        division(temp_list[0], temp_list[1])
    else:
        print("Error Code: Not enough values entered")

for i in out_list:
    print(f"Error Code: {i}")
```

```python
cube = lambda x : x ** 3 # complete the lambda function 
fibo_list = [0]

def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [1]
    elif n > 1:
        fibo_list.append(1)
        for i in range(n - 2):
            fibo_list.append(fibo_list[-1] + fibo_list[-2])
        return fibo_list    
                    
n = int(input())
print(list(map(cube, fibonacci(n))))
```
